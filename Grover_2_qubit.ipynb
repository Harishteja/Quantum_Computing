```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
import numpy as np

qc = QuantumCircuit(2)
# apply hadamard to all qubits 
qc.h([0, 1])
qc.barrier()
qc.x(0)
qc.cz(0, 1)
qc.x(0)
qc.barrier()
# apply hadamard to all qubits
qc.h([0, 1])
# x on all qubits
qc.x([0, 1])
# CZ on all qubits
qc.cz(0, 1)
# x on all qubits
qc.x([0, 1])
# apply hadamard to all qubits
qc.h([0, 1])

qc.measure_all()
print(qc.draw('text'))
#simulate with AerSimulator, 1024 shots
simulator = AerSimulator()
compiled_circuit = transpile(qc, simulator)
result = simulator.run(compiled_circuit, shots=1024).result()
# get counts
counts = result.get_counts(compiled_circuit)
print(counts)
```
