```python

import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator

# --------------------------------------------------
# STEP 1: Calculate the QSP phase angles
# --------------------------------------------------

gamma = 0.5 * np.arccos((np.sqrt(3) - 1) / 2)

phi0 = np.pi/4 - gamma
phi1 = gamma
phi2 = gamma
phi3 = -np.pi/4 - gamma

phis = [phi0, phi1, phi2, phi3]

print("QSP phases (radians):")
print(np.round(phis, 6))


# --------------------------------------------------
# STEP 2: Choose a signal value
# --------------------------------------------------

a = 0.8

theta = -2 * np.arccos(a)

print("\nSignal rotation angle:", theta)


# --------------------------------------------------
# STEP 3: Create the one-qubit QSP circuit
# --------------------------------------------------

qc = QuantumCircuit(1)

# Prepare |+> = H|0>
qc.h(0)

# Apply QSP operations from RIGHT to LEFT

# S(phi3)
qc.rz(-2 * phi3, 0)

# W_X(a)
qc.rx(theta, 0)

# S(phi2)
qc.rz(-2 * phi2, 0)

# W_X(a)
qc.rx(theta, 0)

# S(phi1)
qc.rz(-2 * phi1, 0)

# W_X(a)
qc.rx(theta, 0)

# S(phi0)
qc.rz(-2 * phi0, 0)

# Apply H to extract <+| component
qc.h(0)

print("\nQSP Circuit:")
print(qc.draw("text"))


# --------------------------------------------------
# STEP 4: Simulate the circuit
# --------------------------------------------------

initial_state = Statevector.from_label('0')

final_state = initial_state.evolve(qc)

print("\nFinal statevector:")
print(np.round(final_state.data, 6))


# --------------------------------------------------
# STEP 5: Extract polynomial amplitude
# --------------------------------------------------

qsp_amplitude = final_state.data[0]

target_value = a**3

print("\nQSP amplitude:", np.round(qsp_amplitude, 6))
print("Target a^3:", np.round(target_value, 6))

print("Correct:",
      np.allclose(qsp_amplitude, target_value))


# --------------------------------------------------
# STEP 6: Calculate measurement probability
# --------------------------------------------------

success_probability = np.abs(qsp_amplitude)**2

print("\nProbability of measuring |0>:",
      success_probability)
```
