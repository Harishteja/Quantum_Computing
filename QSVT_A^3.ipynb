```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator

## Define the target matrix

X= np.array([[0, 1], [1, 0]])
Z= np.array([[1, 0], [0, -1]])
A =(1/3)*X + (2/3)*Z
print("Original_matrix", A)

# Construct our LCU Block encoding

# PREPARE /0> = sqrt(1/3)|0> + sqrt(2/3)|1>

theta = 2*np.arcsin(np.sqrt(2/3))
def apppend_block_encoding(qc):

    # PREPARE
    qc.ry(theta, 0)

    # SELECT: apply X if ancilla is 0
    qc.x(0)
    qc.cx(0, 1)
    qc.x(0)

    # SELECT: apply Z if ancilla is 1
    qc.cz(0, 1)

    # PREPARE dagger
    qc.ry(-theta, 0)

# Verify the block encoding

be = QuantumCircuit(2)
apppend_block_encoding(be)
U_A = Operator(be).data

indices = [0, 2] ## little endian ordering(q0 =0,2)
A_encoded = U_A[np.ix_(indices, indices)]
print("Encoded_matrix", A_encoded)
print("Block encoding correct:",
      np.allclose(A_encoded, A))
```
