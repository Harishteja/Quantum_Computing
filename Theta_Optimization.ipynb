```python
import qiskit 
from qiskit import QuantumCircuit,transpile
from qiskit_aer import AerSimulator
from qiskit.circuit import Parameter
import numpy as np
from scipy.optimize import minimize
theta = Parameter('θ')
qc = QuantumCircuit(1, 1)
qc.ry(theta, 0)
qc.measure(0, 0)
sim = AerSimulator()

def cost_function(parameters):
    bound_circuit = qc.assign_parameters({theta: parameters[0]})
    compiled_circuit = transpile(bound_circuit, sim)
    result = sim.run(compiled_circuit, shots=1024).result()
    counts = result.get_counts()
    p1 = counts.get('1',0)/1024
    return 1-p1

intial_guess = [0.2]
result = minimize(cost_function, intial_guess, method = 'COBYLA')

print(f"Optimal θ:     {result.x[0]:.4f}")
print(f"Expected θ=π:  {np.pi:.4f}")
print(f"Final cost:    {result.fun:.4f}")
print(f"P(|1⟩):        {1 - result.fun:.4f}")

```
