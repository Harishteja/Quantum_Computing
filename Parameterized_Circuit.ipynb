```python
import qiskit
from qiskit import QuantumCircuit,transpile 
from qiskit.circuit import Parameter
from qiskit_aer import AerSimulator
import numpy as np

theta = Parameter('θ')

qc = QuantumCircuit(1, 1)
qc.rx(theta, 0)
qc.measure(0, 0)
print(qc.draw('text'))

angles = [0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi]
angle_names = ['0', 'π/4', 'π/2', '3π/4', 'π']
simulator = AerSimulator()
print("\nθ\t\tP(|1⟩)")

for angle, name in zip(angles, angle_names):
    bound_circuit = qc.assign_parameters({theta: angle})
    compilation = transpile(bound_circuit,simulator)
    job = simulator.run(compilation, shots=1000)
    result = job.result()
    counts = result.get_counts()
    print(f"{name}\t\t{counts.get('1', 0)/1000:.3f}")


```
