```python
import qiskit
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit


# Bell state ∥Φ+⟩  = (|00⟩ + |11⟩) / √2
qc = QuantumCircuit(2,2)
qc.h(0)
qc.cx(0, 1)

qc.measure([0,1], [0,1])
#qc.measure_all() if you didn't specify classical bits
print(qc.draw('text'))
from qiskit_aer import AerSimulator
from qiskit import transpile

simulator = AerSimulator()
compiled_circuit = transpile(qc, simulator) #transpile the circuit for the simulator
result = simulator.run(compiled_circuit, shots=1024).result()
print(result.get_counts())

# Bell state ∥Φ-⟩  = (|00⟩ - |11⟩) / √2
qc = QuantumCircuit(2,2)
qc.h(0)
qc.cx(0, 1)
qc.z(0) # Apply Z gate to the first qubit to create the Φ- state    
qc.measure([0,1], [0,1])
compiled_circuit = transpile(qc, simulator)
result = simulator.run(compiled_circuit, shots=1024).result()
print(qc.draw('text'))
print(result.get_counts())  


# Bell state ∥Ψ+⟩  = (|01⟩ + |10⟩) / √2
qc = QuantumCircuit(2,2)    
qc.h(0)
qc.cx(0, 1)
qc.x(0) # Apply X gate to the first qubit to create the Ψ+ state
qc.measure([0,1], [0,1])
compiled_circuit = transpile(qc, simulator)
result = simulator.run(compiled_circuit, shots=1024).result()
print(qc.draw('text'))
print(result.get_counts())      


# Bell state ∥Ψ-⟩  = (|01⟩ - |10⟩) / √2
qc = QuantumCircuit(2,2)
qc.h(0)
qc.cx(0, 1)
qc.x(0) # Apply X gate to the first qubit to create the Ψ- state
qc.z(0) # Apply Z gate to the first qubit to create the Ψ- state
qc.measure([0,1], [0,1])
compiled_circuit = transpile(qc, simulator)
result = simulator.run(compiled_circuit, shots=1024).result()
print(qc.draw('text'))
print(result.get_counts())      
```
